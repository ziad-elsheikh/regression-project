## predicting Livestock Growth Rates: A Custom Regression Analysis on Feed and Health Data

### Data Importing and Preprocessing

```python
import pandas as pd
```

#### loading first dataset and do some configurations


```python
file_path = 'data/raw data/Sires Selection and Resilience Trial Performance and Efficiency Data.xlsx'
df_performance = pd.read_excel(file_path, sheet_name='Insentech_RFI')
```

```python
# remove any additional spaces
df_performance.columns = df_performance.columns.str.strip()

# remove unwanted data
df_performance_cleaned = df_performance.iloc[:, :15].rename(columns={'Calf': 'CalfID'})

# print the dataset
df_performance_cleaned.head()
```

#### loading second dataset and do some configurations

```python
file_path = 'data/raw data/USDA Grant Steers - Master Immunity Spreadsheet.xlsx'
df_immunity = pd.read_excel(file_path)
```

```python
# remove any additional spaces
df_immunity.columns = df_immunity.columns.str.strip()

# printing the dataset
df_immunity.head()
```

#### merge two datasets in one dataframe and export it to a CSV file.

```python
# To meet the >200 rows requirement and preserve valuable physiological data (ODMI, MBW, ADG), 
# a Left Join was utilized instead of an Inner Join.
# This resulted in retaining all 230 calves. 
# The 70 calves lacking hematological data (Cortisol, NLR) introduced missing values (NaNs).
df_merged = pd.merge(df_performance_cleaned, df_immunity, on='CalfID', how='left')

# export new merged data to csv file
df_merged.to_csv(r'data/uncleaned data/Livestock_Data.csv', index=False, encoding='utf-8-sig')
```

```python
print(f"shape of data before merging (performance): {df_performance_cleaned.shape}")
print(f"shape of data before merging (immunity)   : {df_immunity.shape}")
print(f"shape of data after merging               : {df_merged.shape}")

display(df_merged.head())
```
